```python
# Pandas es una libreria especializada en el analisis,manipulación y limpieza de datos estructurados, como los que se vienen en archivos .csv,Excel,bases de datos,etc.
# Es la herramienta mas usada para el análisis de datos.
#Facilita tareas que antes eran tediosas: limpieza,filtrado,ordenamiento,agrupación,calculo de promedios,conteos,etc.

# ¿Que podemos hacer con Pandas?
# Leer archivos y transformarlos en tablas.
# Explorar datos ( ver las primeras fils,tipos,valores faltantes)
# Filtrar y seleccionar columnas o filas.
#Ordenas,agrupar y resumir información.
#Limpiar datos (quitar duplicados,tratar nulos,cambiar formatos)
#Crear columnas nuevas,combinar tablas,aplicas funciones.
```

```python
import pandas as pd
```

```python
#Una series es una estructura de datos unidimensional de Pandas, similar a una lista o un arreglo de Numpy, pero con etiquetas (indices) para cada elemento.
#Es ideal para representar una columna de datos, edad, estatura. etc.
```

```python
edades = pd.Series([25,30,28,22,35])
print(edades)
```

```python

# Indices en  Series de Pandas
edades = pd.Series([25,30,28,22,35], index=["Starleen", "Liza","Steven", "Amparito", "Maria F"])
print(edades)
```

```python
print(edades["Steven"])
#
```

```python
print(edades[2])
#
```

```python
print(edades[0:3])
#
```

```python
print(edades[3])
```

## QUE ES UN DATA FRAME

```python
# Es una estructura que parece una tabla de datos que contiene filas e indices, y columnas con nombre, es la estructura principal de trabajo en pandas

#Crear un dataframe desde un diccionario.

# Explicación: Cada  clave del diccionario es una columna.
#Cada valor (lista) representa los datos de esa  columna
# El dataframe organiza todo en una tabla ordenada.
```

```python
# Crear un diccionario
datos = {
    # Clave  = Valor
    "Nombre": ["Starleen", "Liza","Steven", "Amparito", "Maria F"],
    "Edad": [25,30,28,22,35],
    "Ciudad": ["Medellin", "Bogota", "Bucaramanda", "Bogota", "Cali"]
}
```

```python
df= pd.DataFrame(datos)  # Convierte el diccionario en un dataframe
print(df)
```
