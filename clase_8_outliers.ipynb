```python
import pandas as pd
import numpy as np

#Diccionario con datos mixtos
datos = {
"Cliente": [f"CL{1:03}" for i in range(1, 51)],
"Consumo_kwh": np.append(np.random.normal (350, 40, 45).round(1), [200, 215, 700, 720, 750])
}

df = pd.DataFrame(datos)
df
```

```python
df
```

```python
df.info()
```

```python
df.describe()
```

```python
# Calcular los cuartiles para aplicar el metodo IQR para identificaciónde datos atipicos.
q1 = df['Consumo_kwh'].quantile(0.25)
q3 = df['Consumo_kwh'].quantile(0.75)

IQR = q3 - q1
IQR
```

```python
# Limites para detectar Outliers
limite_inferior = q1 - 1.5 * IQR
limite_superior = q3 + 1.5 * IQR

print(f"Limite Inferior: {limite_inferior}")
print(f"Limite Superior: {limite_superior}")
print("Q1:", q1)
print("Q3:", q3)
#
```

```python
# DETECTAR VALORES FUERA DE RANGO
Outliers = df[(df['Consumo_kwh'] < limite_inferior) | (df['Consumo_kwh'] > limite_superior)]
print("Cantidad de Outliers:")
Outliers
```

```python
# No se recomienda realizar la eliminación de los datos, ya que supera el 5% del total de datos de dataframe. Por ende, vamos a remplazar los datos de cada columna con la mediana,media y moda.

# Vamos a remplazar la columna municipio con la moda.
df_criminal['municipio'].fillna(df_criminal['municipio'].mode()[0], inplace=True)

df_criminal.info()
```

```python
# vamos a remplazar los valores nullos de la columna arma_utilizada con la moda
df_criminal['arma_utilizada'].fillna(df_criminal['arma_utilizada'].mode()[0], inplace=True)
df_criminal.info()
```

```python
# Primero convertimos 'numero_victimas' a tipo numérico para poder calcular el promedio
df_criminal['numero_victimas'] = pd.to_numeric(df_criminal['numero_victimas'], errors='coerce')

# Ahora reemplazamos los valores nulos con el promedio de cada columna
df_criminal['tiempo_respuesta'].fillna(df_criminal['tiempo_respuesta'].mean(), inplace=True)
df_criminal['numero_victimas'].fillna(df_criminal['numero_victimas'].mean(), inplace=True)

df_criminal.info()
```

```python
# Vamos a tratar los valores nulos de la columna edad victima con la moda
df_criminal['edad_victima'].fillna(df_criminal['edad_victima'].mode()[0], inplace=True)
df_criminal.info()
```

```python
# vamos a tratar los datos nulos de la columna tipo_delito con la moda (ya que es categórica)
df_criminal['tipo_delito'] = df_criminal['tipo_delito'].fillna(df_criminal['tipo_delito'].mode()[0])
df_criminal.info()
```

```python
# vamos a tratar los datos nulos de la columna fecha con la moda
df_criminal['fecha'] = df_criminal['fecha'].fillna(df_criminal['fecha'].mode()[0])
df_criminal.info()
#
```

```python
# vamos a tratar los datos nulos de la columna estrato con la moda
df_criminal['estrato'] = df_criminal['estrato'].fillna(df_criminal['estrato'].mode()[0])
df_criminal.info()
#
```

```python
df_criminal.isnull().sum()
```

```python
# Vamos a realizar el manejo de datos atipicos
## Calcular los cuartiles para aplicar el metodo IQR para identificaciónde datos atipicos para la columna número de victimas.
q1 = df_criminal['numero_victimas'].quantile(0.25)
q3 = df_criminal['numero_victimas'].quantile(0.75)

IQR = q3 - q1
IQR
```

```python
# Limites para detectar Outliers
limite_inferior = q1 - 1.5 * IQR
limite_superior = q3 + 1.5 * IQR

print(f"Limite Inferior: {limite_inferior}")
print(f"Limite Superior: {limite_superior}")
print("Q1:", q1)
print("Q3:", q3)
#
```

```python
# DETECTAR VALORES FUERA DE RANGO
Outliers = df_criminal[(df_criminal['numero_victimas'] < limite_inferior) | (df_criminal['numero_victimas'] > limite_superior)]
print("Cantidad de Outliers:")
Outliers

#NO SE PRESENTARON  VALORES ATIPICOS PARA LA COLUMNA NÚMERO DE VICTIMAS. TODOS SE ENCUENTRAN DENTRO DE LOS LIMITES INFERIORES Y SUPERIORES.
# NOTA: ESTO PUEDE ESTAR ASOCIADO A QUE SE REALIZO EL TRATAMIENTO DE LOS VALORES NULOS, LOS CUALES SE ENCUENTRAN DENTRO DEL RANGO POR LA MODA, MEDIANA O MEDIA.
```
