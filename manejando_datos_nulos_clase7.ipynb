##EJERCICIO LIMPIEZA DE DATOS

```python
import numpy as np
```

```python
# Creación del diccionario con datos
datos_energia = {
    "Ciudad": ["Bogotá", "Medellín", "Cali", "Barranquilla", "Bogotá", "Cali", "Medellín", np.nan],
    "Consumo_kWh": [350, 420, np.nan, 390, 310, 405, np.nan, 450],
    "Estrato": [3, 4, 3, np.nan, 2, 3, 4, 2],
    "Subsidio": [True, False, True, True, np.nan, True, False, True]
}

# Creación del DataFrame
df = pd.DataFrame(datos_energia)
```

```python
df
```

```python
#Paso 1: Detección y conteo
#Instrucción: Usa .isnull() y .sum() para contar nulos por columna.
#Pregunta a responder: ¿Cuántos valores faltantes hay en total?

total_nulos= df.isnull().sum()
total_nulos
```

```python
total_nulos.sum()
#Existen 5 valores nulos en total
```

```python
#Paso 2: Eliminación
#Instrucción: Crea una copia del DataFrame y elimina:
#Las filas que tengan al menos un NaN.
#Las columnas que tengan al menos un NaN.
#Pregunta a responder: ¿Cuánta información se pierde en cada caso?

df_sin_filas_nulas = df.dropna()
df_sin_filas_nulas

#Se pierden 5 filas
```

```python
#Eliminación de columnas con datos nulos
df_sin_columnas_nulas = df.dropna(axis=1)
df_sin_columnas_nulas

#Se pierden todas las columnas
```

```python
#Paso 3: Imputación
#Instrucción: Imputa los valores faltantes de estas formas:
#Consumo_kWh: reemplazar con la media.
df["Consumo_kWh"] = df["Consumo_kWh"].fillna(df["Consumo_kWh"].mean())
df
```

```python
#Paso 3: Imputación
#Estrato: reemplazar con la mediana.
df["Estrato"] = df["Estrato"].fillna(df["Estrato"].median())
df
```

```python
#Ciudad: reemplazar con la moda.
df["Ciudad"] = df["Ciudad"].fillna(df["Ciudad"].mode()[0])
df
```

```python
#Subsidio: reemplazar con un valor fijo (False).
df["Subsidio"] = df["Subsidio"].fillna(False)
df
```
