```python
#CREAR UNA NUEVA COLUMNA PARAEL CALCULO TOTAL DE LOS TIQUETES PARA 2 PERSONAS
df_precios["Valor_Total_Ida"] = df_precios["precios ida"] * 2
df_precios
```

## REALIZAR CALCULOS ESTADISTICOS

```python
# Vamos a relaizar calculos estadisticos: Media,Mediana,Moda
print("La Media de los precios ida es: ", df_precios["precios ida"].mean())
print("La Mediana de los precios ida es: ", df_precios["precios ida"].median())
print("La Moda de los precios ida es: ", df_precios["precios ida"].mode())
print("El precio mimino es:", df_precios["precios ida"].min())
print("El precio maximo es:", df_precios["precios ida"].max())
print("La desviacion estandar es:", df_precios["precios ida"].std())
print("La varianza es:", df_precios["precios ida"].var())
print("El rango de los datos es:", df_precios["precios ida"].max() - df_precios["precios ida"].min())
print("La curtuosis es:", df_precios["precios ida"].kurtosis())
print("La asimetria es:", df_precios["precios ida"].skew)
print("El sesgo del precio de ida es:", df_precios['precios ida'].skew())
```
