```python
import numpy as np
```

```python
# Creación de matriz de arrays
matriz = np.array([[1,2,3],
                  [4,5,6]])
print(matriz)
```

```python
# atributos de los arrays
print("Numero de dimensiones: ", matriz.ndim)
print("Forma del array: ", matriz.shape)
print("Tamaño del array: ", matriz.size)
print("Tipo de dato del array", matriz.dtype)
#
```

```python
# mostras elementos especificos de una matriz
print(matriz[0,1]) # Fila , # Columna
print(matriz[1,1]) # Fila , # Columna
#
```

```python
# mostrar toda una fila
print(matriz[1])
```

```python
# mostrr una columa completa
print(matriz[:,1])
```

```python
# mostrar la primera fila y las 2 primeras columnas
print(matriz[1,:2])   #  :2 significa las 2 primeras columnas   1 significa la segunda fila
```

```python
# suma de los elementos al interior de una matriz
print(matriz.sum()) # suma todos los elementos.
print(matriz.sum(axis=0)) # suma todos los elementos por columnas
print(matriz.sum(axis=1)) # suma las filas
```

## ACTIVIDADES BASICAS CON ARRAYS

```python
# Análisis de temperaturas con arrays
temperaturas = np.array([29,31,28,30,27,32,33])
print(temperaturas)
```

```python
# muestra la temperatura del tercer día.
print(temperaturas[2])
```

```python
# Calcula la temperatura promedio de la semana
print(temperaturas.mean())
```

```python
# Muestra las temperaturas mayores a 30 grados
mascara = temperaturas > 30
print(temperaturas[mascara])
# [31 32 33]
```

```python
# Suma 1 grado a todas las emperaturas  y muestra el nuevo array
temperaturas = temperaturas + 1
print(temperaturas)
```

## Análisis de  ventas por región Matriz

```python
ventas = np.array([
    [120,135,150,160], # Region del norte
    [100,110,115,130], # Región del sur
    [90,105,100,120] # Región Centro
])

print(ventas)
```

```python
# 1. muestra las ventas del segundo trimestre para todas las regiones
print(ventas[:,1])
```

```python
# Calcula las ventas totales por la región
print(ventas.sum(axis=1))
```

```python
# Calcula las ventas totales por trimestre.
print(ventas.sum(axis=0)) # suma las ventas totales por columnas (trimestres)
```

```python
totales_trimestre = ventas.sum(axis=0)
print(totales_trimestre.argmax()) #Existe una función de NumPy hermana de .max() que, en vez de devolver el valor máximo, te devuelve la posición (índice) donde se encuentra ese máximo. Se llama argmax (la "g" de "argument", literalmente "el índice/argumento que da el máximo").
```
