### Instalar bibliotecas de Python con `pip`

Para instalar bibliotecas de Python en Colab (o en tu entorno local), puedes usar el comando `pip install`. En Colab, a menudo se usa un signo de exclamación `!` al principio para indicar que es un comando de shell.

La sintaxis general es:

```bash
!pip install nombre_de_la_biblioteca
```

Por ejemplo, si quieres instalar la biblioteca `pandas`, usarías `!pip install pandas`.

````python
# Ejemplo: Instalando la biblioteca 'requests'
!pip install requests

print("La biblioteca 'requests' ha sido instalada (o ya estaba instalada) con éxito.")

# Puedes verificar la instalación importándola
import requests
print(f"Versión de requests: {requests.__version__}")
````

Después de ejecutar esta celda, la biblioteca `requests` estará disponible para usarla en tu notebook. Si la biblioteca ya está instalada, `pip` te lo indicará.
