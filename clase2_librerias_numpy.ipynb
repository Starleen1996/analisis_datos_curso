```python
# Vamos a trabjar con NUMPY
import numpy as np
```

```python
lista = [1,2,3,4,5,6]
array = np.array([1,2,3,4,5]) # esto convierte la lista en array
```

```python
print("Lista:", lista)
print("Array:", array)
```

```python
print(type(lista)) ## muestra el tipo de dato
print(type(array))
```

```python
# sumar los elementos con función SUM para un array

print("La suma de los elementos es: " , array.sum())
```

```python
suma_lista = 0
for num in lista:
    suma_lista += num # Esto es para crear la suma de una lista
print("La suma de los elementos de la lista es: ", suma_lista)
```

```python
# Función muliplicación
doble_array = array * 2
print("El doble de los elementos del array es: ", doble_array)
```

```python
# PROPIEDADES DEL ARRAY
print("Dimensión del array: ", array.ndim)
print("Forma del array: ", array.shape)
print("Tamaño del array: ", array.size)
print("Tipo de dato del array: ", array.dtype)
```

```python
# EJERCICIO ARRAY 2
#Ejercicio 2.1
#Explora el array
numeros = np.array([10,20,30,40,50])
print("El array es: ", numeros)
print("Dimensión del array: ", numeros.ndim)
print("Forma del array: ", numeros.shape)
print("Tamaño del array: ", numeros.size)
print("Tipo de dato del array: ", numeros.dtype)
```

```python
# Ejercicio 2.3
# Calcular la suma y el promedio
print("La suma de los elementos del array es: ", numeros.sum())
print("El promedio de los elementos del array es: ", numeros.mean())
```

```python
# Ejercicio 3
# Crear un array precios con
precios = np.array([10000,15000,20000,25000,30000])
```

```python
# 3.1 Calcular el precio con el 10% de aumento
precios_aumento = precios * 1.1
print("El precio con el 10% de aumento es: ", precios_aumento)
```

```python
# otra opción para clacular el 10% de aumento
precios_aumento = precios + (precios * 0.1)
print("El precio con el 10% de aumento es: ", precios_aumento)
```
